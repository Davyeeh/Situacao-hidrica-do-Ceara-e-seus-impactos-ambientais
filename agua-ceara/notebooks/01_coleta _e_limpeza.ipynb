```python
import pandas as pd
import glob
import os

arquivos = glob.glob("./dados tratados/FUNCEME/*.csv")

lista = []

for arquivo in arquivos:
    df_temp = pd.read_csv(arquivo)
    
    ano = os.path.basename(arquivo).split(".")[0]
    df_temp["ano"] = int(ano)
    
    lista.append(df_temp)

df_funceme = pd.concat(lista, ignore_index=True)

print(df_funceme.head())
print(df_funceme.shape)
```
